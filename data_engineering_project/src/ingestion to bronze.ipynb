```sql
%python
dbutils.widgets.text("catalog","dev")

catalog=dbutils.widgets.get("catalog")
```

```sql
create table if not exists identifier(:catalog||'.bronze.suppliers_raw');
create table if not exists identifier(:catalog||'.bronze.sales_raw');
create table if not exists identifier(:catalog||'.bronze.customers_raw');
create table if not exists identifier(:catalog||'.bronze.products_raw');
```

```sql
DECLARE OR REPLACE VARIABLE tbl STRING;
DECLARE OR REPLACE VARIABLE path STRING;

SET VAR tbl = :catalog || '.bronze.suppliers_raw';
SET VAR path = '/Volumes/' || :catalog || '/bronze/raw/suppliers';

EXECUTE IMMEDIATE '
  COPY INTO IDENTIFIER(:tbl)
  FROM :path
  FILEFORMAT = CSV
  FORMAT_OPTIONS ("header" = "true", "inferSchema" = "true")
  COPY_OPTIONS ("mergeSchema" = "true")
'
USING tbl AS tbl, path AS path;
```
